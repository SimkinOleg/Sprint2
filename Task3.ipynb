### Система управления уведомлениями (Вариант 24).

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание: Проект демонстрирует разные виды наследования. Простое: Notification -> SMSNotification. Сложное: Notification -> EmailNotification -> MarketingEmailNotification (маркетинговое письмо). Множественное: так как C# не поддерживает множественное наследование классов, класс PushNotification наследует базовый класс и реализует интерфейс IInteractiveNotification.
#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public interface IInteractiveNotification { void OpenApp(); }

public class Notification
{
    public int NotificationId { get; set; }
    public string MessageText { get; set; }
    public string Type { get; set; }
    protected int Priority { get; set; } // Доп. атрибут

    public Notification(int id, string msg, string type, int priority)
    {
        NotificationId = id; MessageText = msg; Type = type; Priority = priority;
    }

    public virtual void DisplayNotification() => Console.WriteLine($"[{Type}] {MessageText}");
    public virtual void SendNotification() => Console.WriteLine($"Отправка {Type}...");
    public virtual string GetNotificationDetails() => $"Приоритет: {Priority}";
}

// Простое наследование
public class SMSNotification : Notification
{
    public string PhoneNumber { get; set; }
    public SMSNotification(int id, string msg, int priority, string phone) : base(id, msg, "SMS", priority) => PhoneNumber = phone;
    public override void SendNotification() => Console.WriteLine($"[SMS] Отправка на {PhoneNumber}");
}

// Сложное (многоуровневое) наследование
public class EmailNotification : Notification
{
    public string EmailAddress { get; set; }
    public EmailNotification(int id, string msg, int priority, string email) : base(id, msg, "Email", priority) => EmailAddress = email;
    public override void SendNotification() => Console.WriteLine($"[Email] Отправка на {EmailAddress}");
}
public class MarketingEmailNotification : EmailNotification
{
    public string PromoCode { get; set; } // Доп. атрибут
    public MarketingEmailNotification(int id, string msg, int priority, string email, string promo) 
        : base(id, msg, priority, email) => PromoCode = promo;
        
    public override void SendNotification()
    {
        base.SendNotification();
        Console.WriteLine($"  (Промокод: {PromoCode})");
    }
}

// Множественное наследование (Класс + Интерфейс)
public class PushNotification : Notification, IInteractiveNotification
{
    public string Platform { get; set; }
    public PushNotification(int id, string msg, int priority, string platform) : base(id, msg, "Push", priority) => Platform = platform;
    
    public override void DisplayNotification() => Console.WriteLine($"[Push-{Platform}] {MessageText}");
    public void OpenApp() => Console.WriteLine($"Открыто приложение из Push {NotificationId}");
}

// Проверка
var marketing = new MarketingEmailNotification(1, "Скидка 50%", 2, "user@mail.com", "SALE50");
marketing.SendNotification();
var push = new PushNotification(2, "Новое сообщение", 1, "iOS");
push.OpenApp();
```
