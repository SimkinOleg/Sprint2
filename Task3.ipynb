#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

// ================== ИНТЕРФЕЙСЫ (Множественное наследование) ==================
public interface ISwimmable { void Swim(); }
public interface IFlyable { void Fly(); }

// ================== БАЗОВЫЙ КЛАСС (Простое наследование) ==================
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    protected int Energy; // protected = доступно наследникам

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
        Energy = 100;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Животное: {Name}, Возраст: {Age}, Энергия: {Energy}");
    }

    public void Eat(int amount)
    {
        Energy += amount;
        Console.WriteLine($"{Name} поел. Энергия: {Energy}");
    }
}

// ================== СЛОЖНОЕ НАСЛЕДОВАНИЕ ==================
// Уровень 1: Animal -> Уровень 2: Mammal -> Уровень 3: Lion
public class Mammal : Animal
{
    public bool HasFur { get; set; } // Атрибут млекопитающего

    public Mammal(string name, int age, bool hasFur) : base(name, age)
    {
        HasFur = hasFur;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  [Млекопитающее] Шерсть: {HasFur}");
    }
}

public class Lion : Mammal
{
    public int PrideSize { get; set; } // Размер прайда

    public Lion(string name, int age, bool hasFur, int prideSize) : base(name, age, hasFur)
    {
        PrideSize = prideSize;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"  [Лев] Размер прайда: {PrideSize}");
    }

    public void Roar()
    {
        Console.WriteLine($"{Name} рычит: РРРРР!");
    }
}

// ================== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (Интерфейсы) ==================
// Пингвин наследует Animal (простое) и реализует ISwimmable
public class Penguin : Animal, ISwimmable
{
    public double SwimSpeed { get; set; } // Атрибут пингвина

    public Penguin(string name, int age, double speed) : base(name, age)
    {
        SwimSpeed = speed;
    }

    public void Swim()
    {
        Energy -= 10;
        Console.WriteLine($"{Name} плывет со скоростью {SwimSpeed} км/ч. Энергия: {Energy}");
    }
}

// Орел наследует Animal (простое) и реализует IFlyable
public class Eagle : Animal, IFlyable
{
    public double WingSpan { get; set; } // Атрибут орла

    public Eagle(string name, int age, double wingSpan) : base(name, age)
    {
        WingSpan = wingSpan;
    }

    public void Fly()
    {
        Energy -= 20;
        Console.WriteLine($"{Name} летит. Размах крыльев: {WingSpan} м. Энергия: {Energy}");
    }
}

// ... (все твои классы Animal, Lion, Penguin, Eagle и интерфейсы идут выше) ...

// === ОБЕРТКА ДЛЯ ПРОВЕРКИ ===
public class Program
{
    public static void Main()
    {
        Console.WriteLine("=== ЗООПАРК ===");

        // 1. Лев (Сложное наследование: Animal -> Mammal -> Lion)
        Lion lion = new Lion("Симба", 5, true, 10);
        lion.DisplayInfo();
        lion.Roar();
        lion.Eat(20);

        Console.WriteLine();

        // 2. Пингвин (Множественное: Animal + ISwimmable)
        Penguin penguin = new Penguin("Пингвинчик", 2, 15.5);
        penguin.DisplayInfo();
        penguin.Swim();

        Console.WriteLine();

        // 3. Орел (Множественное: Animal + IFlyable)
        Eagle eagle = new Eagle("Орлуша", 3, 2.5);
        eagle.DisplayInfo();
        eagle.Fly();

        // 4. Демонстрация полиморфизма
        Console.WriteLine("\n=== Кормление всех животных ===");
        Animal[] allAnimals = { lion, penguin, eagle };
        foreach (Animal a in allAnimals)
        {
            a.Eat(10);
        }
    }
}
```
